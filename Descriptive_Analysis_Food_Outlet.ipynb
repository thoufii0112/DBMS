Title: Descriptive Analysis of a Food Outlet Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("HyderabadResturants.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df.head()
```

```python
print("Average Customer Rating:", pd.to_numeric(df["ratings"], errors="coerce").mean())
```

```python
highest = df.loc[pd.to_numeric(df["ratings"], errors="coerce").idxmax()]
print("Restaurant with Highest Rating:")
print("Name:", highest["names"])
print("Rating:", highest["ratings"])
```

```python
print("Average Price for One Person:", df["price for one"].mean())
```

```python
top10 = df.sort_values(by="ratings", ascending=False).head(10)
```

```python
print("Top 10 Restaurants with the Highest Rating")
print(top10[["names", "ratings"]])
```

```python
print("Top 10 Most Common Cuisines")
print(df["cuisine"].value_counts().head(10))
```

```python
top10 = df.sort_values(by="ratings", ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top10["names"], top10["ratings"])
plt.title("Top 10 Highest Rated Restaurants")
plt.xlabel("Restaurant Name")
plt.ylabel("Rating")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["ratings"], bins=10)
plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Ratings")
plt.ylabel("Number of Restaurants")
plt.show()
```

```python
top_cuisine = df["cuisine"].value_counts().head(10)

plt.figure(figsize=(8,8))
plt.pie(top_cuisine.values,
        labels=top_cuisine.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Top 10 Most Common Cuisines")
plt.show()
```

Descriptive Analysis Performed
The following business questions were analyzed using the Hyderabad Restaurants dataset:

What is the average customer rating?
Which restaurant has the highest customer rating?
What is the average price for one person?
Which are the top 10 restaurants with the highest ratings?
What are the top 10 most common cuisines?
Data Visualizations Created
The following charts were created to support the analysis:

Bar Chart – Top 10 Highest Rated Restaurants
Histogram – Distribution of Restaurant Ratings
Pie Chart – Top 10 Most Common Cuisines
Conclusion
The descriptive analysis of the Hyderabad Restaurants dataset shows that the average customer rating is approximately 3.95, indicating that most restaurants provide good service and customer satisfaction. The analysis also identified the highest-rated restaurants, the average price for one person, and the most popular cuisines among customers. These insights can help restaurant owners understand customer preferences, improve their services, and make better business decisions.
